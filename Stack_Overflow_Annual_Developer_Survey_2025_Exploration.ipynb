# **Most Appropriate Dataset is: Stack Overflow Annual Developer Survey 2025**

**Loading**

```python
import kagglehub, os, pandas as pd

path = kagglehub.dataset_download("edoardogalli/stack-overflow-annual-developer-survey-2025")
print(os.listdir(path))

df = pd.read_csv(f"{path}/survey_results_public.csv", low_memory=False)
```

**Number of Rows and Columns**


```python
rows = df.shape[0]
columns = df.shape[1]
print(f'Rows: {rows}')
print(f'Columns: {columns}')
```

**Datatypes Contained in the Dataset**



```python
print(df.dtypes.value_counts())
df.dtypes
```

**Dataset Complete?**

```python
print(f"Any missing values? {df.isna().any().any()}")
```
